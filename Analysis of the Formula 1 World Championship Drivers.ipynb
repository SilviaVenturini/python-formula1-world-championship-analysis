```python
import csv

def analyze_driver_performance(pilot_name):

  """
  Receives the name of a driver and returns a list composed of:
  [total_points, total_wins, total_podiums], namely:
  - total_points = sum of the points from all races for the driver
  - total_wins = total number of races won (finished in first place)
  - total_podiums = total number of times the driver finished in the top three positions

  """
  # Podium position mapping with the corresponding score
  position_score_dict = {0:0, 1:10, 2:8, 3:6, 4:5, 5:4, 6:3, 7:2, 8:1}

  with open("formula1_data.csv", "r", encoding='utf-8') as file_csv:

    total_points = 0
    total_wins = 0
    total_podiums = 0
    pilot_is_found = False # Indicates whether the driver is present in the CSV file
    score_list = []

    # Iterates over the rows of the CSV file "formula1_data.csv" opened as a dictionary
    for row in csv.DictReader(file_csv):
      if row["Driver"] == pilot_name:
        pilot_is_found = True
        position = int(row["Position"]) # Identifies the driver's finishing position
        total_points += position_score_dict[position] # Total sum of points
        if position == 1:
          total_wins +=1 # Total number of wins (first place finishes)
        if position <= 3 and position > 0:
          total_podiums +=1 # Total number of podiums (top three finishes)

    if pilot_is_found == False: # If the driver's name is not present in the CSV file
      print(f"Il pilota {pilot_name} non è presente nel file.") # Prints the message and returns an empty list
    else:
      score_list = [total_points, total_wins, total_podiums] # Creates the output list

    return score_list

# Script for functionality testing
pilot_score_list = analyze_driver_performance("Massa")
print(pilot_score_list)
```

```python
import csv

def generate_final_standings():

  """
  Generates and writes a txt file containing the drivers' names and their total scores.

  """

  with open("formula1_data.csv", "r", encoding='utf-8') as file_csv:

      driver_dict = {} # Dictionary with driver names and their corresponding scores

      # Iterates over the rows of the CSV file "formula1_data.csv" opened as a dictionary
      for row in csv.DictReader(file_csv):
        pilot = row["Driver"]
        if pilot not in driver_dict:
          driver_dict[pilot] = 0 # Initializes the dictionary by inserting driver names without duplicates

  # Inserts the total score for each driver into the dictionary using the previous function
  for key in driver_dict.keys():
    player_scores = analyze_driver_performance(key)
    driver_dict[key] = player_scores[0] # Individual driver's total score

  # Creates the file "Drivers_Standings_2008.txt", writes to it, and closes it
  with open("Drivers_Standings_2008.txt", "w", encoding='utf-8') as file_txt:
    file_txt.write("Drivers Standings 2008 Formula 1\n") # File header
    for key in driver_dict.keys():
      file_txt.writelines(str(key) + ": " + str(driver_dict[key])+"\n")

# Script for functionality testing
generate_final_standings()
```

```python
import csv

def constructors_standings():

  """
  Generates a dictionary composed of:
  {team:final_score}, namely:
  - team = car manufacturer / racing team
  - final_score = sum of the points of all drivers who raced for that team

  """

  formula1_data = open("formula1_data.csv") # Opens the file "formula1_data.csv"

  with open("Drivers_Standings_2008.txt", "r", encoding='utf-8') as file_txt:

      # Reads all lines from the txt file except the header
      file_txt_list = file_txt.readlines()
      name_score_team_list = file_txt_list[1:]

      # Creates a list of lists called "name_score_team_list"
      # that associates the driver's name with their score
      index = 0
      for element_list in name_score_team_list:
        name_score_team_list[index]= element_list.split()
        name_score_team_list[index][0] = name_score_team_list[index][0].strip(":")
        index+=1

      # Searches for and adds to the list the team name associated with the driver
      for name_score_list in name_score_team_list:
        for row_formula1_data in csv.DictReader(formula1_data):
          if row_formula1_data["Driver"] == name_score_list[0]:
            name_score_list.append(row_formula1_data["Team"])
            formula1_data.seek(0)
            break

  formula1_data.close() # Manually closes the CSV file

  # Creates a list containing team names ordered according to the drivers,
  # i.e., a copy of the elements at index 2 of "name_score_team_list""
  driver_team_list = []
  for teams in name_score_team_list:
    driver_team_list.append(teams[2])

  # Creates a list of team names without duplicates
  list_single_team = list(set(driver_team_list))

  # Initializes a dictionary (output of the current function) to store each team's total score
  teams_dict = {}
  # Iterates over each team in the list
  for single_team in list_single_team:
      team_score = 0  # Initializes the team score to 0 for each new team

      # Iterates through all teams in the "driver_team_list", getting index and team name
      for i, team in enumerate(driver_team_list):
          # Checks if the current team matches the team being examined
          if team == single_team:
              # Adds the score of the driver corresponding to the team
              # name_score_team_list[i][1] contains the driver's points
              team_score += int(name_score_team_list[i][1])
          # Inserts the team name (key) and its score (value) into the dictionary
          teams_dict[single_team] = team_score
  return teams_dict

# Script for functionality testing
constructors_scores = constructors_standings()
print(constructors_scores)
```
